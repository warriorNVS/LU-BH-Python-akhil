```python
# 1. Start of day: the checklist
checklist = ["Do laundry", "Study for 1 hour", "Call mom", "Go to the gym"]

# Two empty piles
completed_tasks = []
incomplete_tasks = []

# 2. End of day: go through each task and ask "did I finish it?"
for task in checklist:
    answer = input(f"Did you finish '{task}'? (yes/no): ")
    if answer.lower() == "yes":
        completed_tasks.append(task)      # move to completed
    else:
        incomplete_tasks.append(task)     # move to incomplete

# 3. Show the summary
print("Completed:", completed_tasks)
print("Still to do:", incomplete_tasks)
```
